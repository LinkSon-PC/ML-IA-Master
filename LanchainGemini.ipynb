## Conectar LangChain con Gemini

````python
from dotenv import load_dotenv
import os

from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

api_key = os.getenv("GOOGLE_API_KEY")

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=api_key,
    temperature=0.3,
    max_tokens=1000,
    timeout=30
)

response = llm.invoke("Qué es LangChain")

print(response.content)
````

## Crear tu primera Tool

Aquí empieza el poder de LangChain.

````python
from langchain.tools import tool

@tool("calculator", description="Performs arithmetic calculations. Use this for any math problems.")
def calc(expression: str) -> str:
    """Evaluate mathematical expressions."""
    return str(eval(expression))
````

````python
from langchain.agents import create_agent

tools = [calc]

agent = create_agent(
    tools=tools,
    model=llm,
    name="LangChain_Gemini_Assistant"
)
````

````python

response = agent.invoke({"messages":["Calculate (500 * 10) + 250"]})

for m in response["messages"]:
    m.pretty_print()
````

## Memoria

- Cada usuario tiene un session_id
- Las conversaciones se guardan por sesión
- Gemini puede recordar contexto anterior

````python
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.messages import HumanMessage
````

````python
# =========================
# ALMACÉN DE SESIONES
# =========================

store = {}
````

````python
# =========================
# OBTENER MEMORIA POR SESSION ID
# =========================

def get_session_history(session_id: str):

    if session_id not in store:

        store[session_id] = InMemoryChatMessageHistory()

    return store[session_id]

# =========================
# WRAPPER CON MEMORIA
# =========================

chain_with_history = RunnableWithMessageHistory(
    agent,
    get_session_history,
    input_messages_key="input",
    history_messages_key="chat_history"
)

# =========================
# SESSION ID
# =========================

session_id = "usuario_001"
````

````python
# =========================
# PRIMER MENSAJE
# =========================

response1 = chain_with_history.invoke(
    {
        "input": "Hola, me llamo Anthony"
    },
    config={
        "configurable": {
            "session_id": session_id
        }
    }
)

print("AI:", response1.content)

# =========================
# SEGUNDO MENSAJE
# =========================

response2 = chain_with_history.invoke(
    HumanMessage("Cómo me llamo?"),
    config={
        "configurable": {
            "session_id": session_id
        }
    }
)

print("AI:", response2.content)


# =========================
# TERCER MENSAJE
# =========================

response3 = chain_with_history.invoke(
    HumanMessage("calcula 150 * 20"),
    config={
        "configurable": {
            "session_id": session_id
        }
    }
)

print("AI:", response3.content)

# =========================
# CUARTO MENSAJE
# =========================

response4 = chain_with_history.invoke(
    HumanMessage("El resultado anterior es par o impar"),
    config={
        "configurable": {
            "session_id": session_id
        }
    }
)

print("AI:", response4.content)
````
