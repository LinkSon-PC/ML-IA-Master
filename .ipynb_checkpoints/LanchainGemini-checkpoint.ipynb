## Conectar LangChain con Gemini

```python
from dotenv import load_dotenv
import os

from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

api_key = os.getenv("GOOGLE_API_KEY")

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=api_key,
    temperature=0.3,
    max_tokens=1000,
    timeout=30
)

response = llm.invoke("Qué es LangChain")

print(response.content)
```

## Crear tu primera Tool

Aquí empieza el poder de LangChain.

```python
from langchain.tools import tool

@tool("calculator", description="Performs arithmetic calculations. Use this for any math problems.")
def calc(expression: str) -> str:
    """Evaluate mathematical expressions."""
    return str(eval(expression))
```

```python
from langchain.agents import create_agent

tools = [calc]

agent = create_agent(
    tools=tools,
    model=llm,
    name="LangChain_Gemini_Assistant"
)

response = agent.invoke({"messages":["Calculate (500 * 10) + 250"]})

for m in response["messages"]:
    m.pretty_print()
```

```python
from langchain_community.tools import DuckDuckGoSearchRun

search_tool = DuckDuckGoSearchRun()

tools = [calc, search_tool]

agent = create_agent(
    tools=tools,
    model=llm,
    name="LangChain_Gemini_Assistant"
)

response = agent.invoke({"messages":["Busca quién ganó el mundial 2022 y calcula 150 * 20"]})

for m in response["messages"]:
    m.pretty_print()
```
